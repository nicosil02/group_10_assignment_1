Create a dictionary containing the population of the following cities

```python
population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}
```

1. Print the complete dictionary

```python
print(population)
```

2. Display only its keys using `keys()`

```python
population.keys()
```

3. Get the population of `"Lima"` using `get()`

```python
population.get('Lima')
```

4. Add `"Cusco": 428000` to the dictionary

```python
population.update( { "Cusco" :  428000 } )
print(population)
```

5. Remove `"Trujillo"` using `pop()`


```python
population.pop("Trujillo")
```

6. Print the final dictionary


```python
print(population)
```
