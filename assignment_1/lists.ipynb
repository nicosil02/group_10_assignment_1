Using this list:

1. Print the original list.
2. Add the grade 17 using append().
3. Sort the grades from lowest to highest using sort().
4. Display:
    - the highest grade;
    - the lowest grade;
    - the total number of grades.

```python
# 1. Print the original list.
print(grades)
```

```python
# 2. Add the grade 17 using append().
grades.append(17)
```

```python
# 3. Sort the grades from lowest to highest using sort().
grades.sort()
```

```python
# 4. Display: the highest grade; the lowest grade; the total number of grades.
print("Highest grade:", max(grades))
print("Lowest grade:", min(grades))
print("Total number of grades:", len(grades))
```
